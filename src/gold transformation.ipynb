```python
df_yellow = spark.sql("""
    select
        id_vendor,
        pickup_datetime as dh_inicio_corrida,
        dropoff_datetime as dh_final_corrida,
        vl_total as vl_total_corrida,
        nr_passenger as nr_passageiros
    from silver.taxidata.yellow_taxi_data
    """)
```

```python
df_yellow.write.format("delta").mode("overwrite").saveAsTable("gold.taxidata.yellow_taxi_data")
```
