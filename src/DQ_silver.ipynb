## Validações de Qualidade de Dados

Validações automatizadas para garantir a integridade dos dados após transformação.

```python
import datetime

# Validações de Qualidade - Yellow Taxi
# 1. Validar valores negativos
negative_values = spark.sql("""
    SELECT 
        COUNT(*) as total_negativos,
        SUM(CASE WHEN vl_total < 0 THEN 1 ELSE 0 END) as vl_total_negativo,
        SUM(CASE WHEN nr_passenger < 0 THEN 1 ELSE 0 END) as passageiros_negativo,
        SUM(CASE WHEN distance < 0 THEN 1 ELSE 0 END) as distancia_negativa
    FROM silver.taxidata.yellow_taxi_data
""").collect()[0]

assert negative_values.vl_total_negativo == 0, f"Encontrados {negative_values.vl_total_negativo} valores negativos em vl_total"
assert negative_values.passageiros_negativo == 0, f"Encontrados {negative_values.passageiros_negativo} valores negativos em nr_passenger"
assert negative_values.distancia_negativa == 0, f"Encontrados {negative_values.distancia_negativa} valores negativos em distance"
print("Sem valores negativos")
```

```python
# 2. Validar período de dados
periodo = spark.sql("""
    SELECT 
        MIN(DATE(pickup_datetime)) as data_minima,
        MAX(DATE(pickup_datetime)) as data_maxima
    FROM silver.taxidata.yellow_taxi_data
""").collect()[0]

assert periodo.data_minima >= datetime.date(2023, 1, 1), f"Data mínima fora do período: {periodo.data_minima}"
assert periodo.data_maxima <= datetime.date(2023, 5, 31), f"Data máxima fora do período: {periodo.data_maxima}"
print(f"Validação 2: Período válido ({periodo.data_minima} a {periodo.data_maxima})")
```

```python
# 3. Validar valores nulos (devem ser zero após dropna)
nulls = spark.sql("""
    SELECT 
        SUM(CASE WHEN id_vendor IS NULL THEN 1 ELSE 0 END) as vendor_null,
        SUM(CASE WHEN vl_total IS NULL THEN 1 ELSE 0 END) as total_null,
        SUM(CASE WHEN nr_passenger IS NULL THEN 1 ELSE 0 END) as passenger_null,
        SUM(CASE WHEN pickup_datetime IS NULL THEN 1 ELSE 0 END) as pickup_null
    FROM silver.taxidata.yellow_taxi_data
""").collect()[0]

assert nulls.vendor_null == 0, f"Encontrados {nulls.vendor_null} nulos em id_vendor"
assert nulls.total_null == 0, f"Encontrados {nulls.total_null} nulos em vl_total"
assert nulls.passenger_null == 0, f"Encontrados {nulls.passenger_null} nulos em nr_passenger"
assert nulls.pickup_null == 0, f"Encontrados {nulls.pickup_null} nulos em pickup_datetime"
print("Validação 3: Sem valores nulos")
```
